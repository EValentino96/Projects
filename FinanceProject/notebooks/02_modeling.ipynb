Doing a linear regression and some forest models as a first pass, using all columns to get a feel of what's important and what's not for our short-term prediction, I expect most short-term indicators to be most valuable with a couple long term indicators helping out, just as it would if a person were doing this prediction. I want to predict each individual 5 days in the future as well as the singular 5th day in the future for comparison.

```python
nvda_for_modeling.columns
```

```python
# Setting up the target columns and feature columns
# Starting by just predicting the 5th day, but eventually will do all 5 days
# nvda["Target_day1"] = nvda["Close"].shift(-1)
# nvda["Target_day2"] = nvda["Close"].shift(-2)
# nvda["Target_day3"] = nvda["Close"].shift(-3)
# nvda["Target_day4"] = nvda["Close"].shift(-4)
nvda_for_modeling["Target_day5"] = nvda_for_modeling["Close"].shift(-5)

feature_cols = ['Close', 'High', 'Low', 'Open', 'Volume', 'MA_14', 'EMA_14',
       'EMA_50', 'Volatility_14', 'Upper_BB_14', 'Lower_BB_14', 'ATR_14',
       'Volatility_50', 'Upper_BB_50', 'Lower_BB_50', 'High-Low', 'High-Close',
       'Low-Close', 'True_Range', 'ATR_50', 'Adaptive_Window', 'Adaptive_MA',
       'Adaptive_Std', 'Adaptive_Upper_BB', 'Adaptive_Lower_BB',
       'Rolling_Max_14', 'Rolling_Min_14', 'Rolling_Max_50', 'Rolling_Min_50',
       'MA_50', 'Close_FFT', 'Doji', 'Engulfing', 'Hammer', 'Doji_Significant',
       'Engulfing_Significant', 'Hammer_Significant', 'ADX_14', 'PLUS_DI_14',
       'MINUS_DI_14', 'ADX_50', 'PLUS_DI_50', 'MINUS_DI_50', 'RSI_14',
       'RSI_50', 'MFI_14', 'MFI_50']
```

```python
nan_rows = nvda_for_modeling[nvda_for_modeling[feature_cols + ["Target_day5"]].isnull().any(axis=1)]
print(nan_rows.index)
```

```python
nvda_for_modeling[feature_cols + ["Target_day5"]].isnull().sum()
```

```python
# Then drop NaNs only in this one
nvda_for_modeling.dropna(subset=feature_cols + ["Target_day5"], inplace=True)

# Define features and target
X = nvda_for_modeling[feature_cols]
y = nvda_for_modeling["Target_day5"]

# Chronological 80/20 split
split_idx = int(len(nvda) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# Train
rf = RandomForestRegressor(random_state=42)
rf.fit(X_train, y_train)

# Predict
preds_rf = rf.predict(X_test)

# Evaluate
mse = mean_squared_error(y_test, preds_rf)
r2 = r2_score(y_test, preds_rf)

print(f"Random Forest MSE: {mse:.2f}")
print(f"Random Forest R²: {r2:.4f}")
```

```python
from xgboost import XGBRegressor

xgb = XGBRegressor(random_state=42)
xgb.fit(X_train, y_train)
preds_xgb = xgb.predict(X_test)

# Same evaluation code
mse = mean_squared_error(y_test, preds_xgb)
r2 = r2_score(y_test, preds_xgb)

print(f"Random Forest MSE: {mse:.2f}")
print(f"Random Forest R²: {r2:.4f}")
```

So this explains it, all our dimensions are correct, so no issues there, but if we look at the training labels and test labels, the average value in the train is much lower than the average of the test labels. Forest algorithms are great at interpolating but fail when trying to extrapolate data, so this is expected. However we can "fix" this by changing how our forest algorithms see the data. if we look at the logged return of the 5 day target, we will have a much more uniform label set.

```python
baseline_preds = X_test["Close"]  # use today's close as the naive guess
print("Naive MSE:", mean_squared_error(y_test, baseline_preds))
print("Naive R²:", r2_score(y_test, baseline_preds))
```

Random forest did much better, we're looking at the logged return MSE and R2, but then we're also converting back to price and we can see how spot on it is, it's much worse in the log scale since all values are smaller and minor differences are heavily penilized.

XGBoost with tuning did great with tuning. Now let's try to individually predict the next 5 days:

```python
# Create the return for the future days: 1 through 5
for i in range(1, 6):
    nvda_for_modeling[f"Target_return_day{i}"] = np.log(nvda_for_modeling["Close"].shift(-i) / nvda_for_modeling["Close"])
```

```python
return_cols = [f"Target_return_day{i}" for i in range(1, 6)]
nvda_clean = nvda_for_modeling.dropna(subset=feature_cols + return_cols)
```

```python
from sklearn.multioutput import MultiOutputRegressor

X = nvda_clean[feature_cols]
y = nvda_clean[return_cols]

split_idx = int(len(X) * 0.8)
X_train, X_test = X.iloc[:split_idx], X.iloc[split_idx:]
y_train, y_test = y.iloc[:split_idx], y.iloc[split_idx:]

base_model_1 = RandomForestRegressor(random_state=42)

base_model_2 = XGBRegressor(
    n_estimators=200,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

multi_model_1 = MultiOutputRegressor(base_model_1)
multi_model_1.fit(X_train, y_train)
preds_1 = multi_model_1.predict(X_test)

multi_model_2 = MultiOutputRegressor(base_model_2)
multi_model_2.fit(X_train, y_train)
preds_2 = multi_model_2.predict(X_test)
```

```python
base_close = nvda_clean["Close"].iloc[split_idx:].values.reshape(-1, 1)
actual_price = np.exp(y_test.values) * base_close

predicted_price_1 = base_close * np.exp(preds_1)
predicted_price_2 = base_close * np.exp(preds_2)

mse_price_1 = mean_squared_error(actual_price, predicted_price_1)
r2_price_1 = r2_score(actual_price, predicted_price_1)

mse_price_2 = mean_squared_error(actual_price, predicted_price_2)
r2_price_2 = r2_score(actual_price, predicted_price_2)
```

```python
plt.figure(figsize=(14, 8))

for i in range(5):
    plt.plot(actual_price[:, i], label=f"Actual Day {i+1}")
    plt.plot(predicted_price_1[:, i], linestyle="--", label=f"Predicted Day {i+1}")

plt.title("Predicted vs Actual Prices by Future Day (Random Forest)")
plt.xlabel("Test Sample Index")
plt.ylabel("Price")
plt.legend(ncol=2)
plt.grid(True)
plt.tight_layout()
plt.show()
```

#### Thoughts

* The actual vs predicted curves are tightly aligned across all 5 future days. That means the model is capturing not just the next-day trend, but momentum and directionality across multiple days.

* Error increases gradually by day: You can see that predicted Day 1 is closest to the actual line, and by Day 5 the deviation grows a little—which is totally expected in multi-day forecasts.

* No wild overshooting or drift -> the model is stable and not hallucinating future spikes.

How this is powerful:

* Strategic decisions (e.g. “wait 3 more days before selling”)

* Confidence analysis (see where the lines are more tightly grouped)

* Trading heuristics
